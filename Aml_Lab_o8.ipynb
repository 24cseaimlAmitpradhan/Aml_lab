```python
import pandas as pd
import numpy as np
data = {
    'x1': [1, 2, 3, 7, 8, 9],
    'x2': [2, 3, 4, 6, 7, 8],
    'x3': [5, 4, 3, 3, 2, 1],
    'y': [0, 0, 0, 1, 1, 1]
}
df = pd.DataFrame(data)
display(df)

def correlation(df):
    return df.corr(method='pearson')
corr = correlation(df)



def merit(corr, features, target):
    k = len(features)
    r_cf = corr.loc[features, target].mean()
    if k == 1:
        r_ff = 0
    else:
        feature_corr = corr.loc[features, features].values
        r_ff = (feature_corr.sum() - k) / (k * (k - 1))
    merit_score = (k * r_cf) / np.sqrt(
        k + k * (k - 1) * r_ff
    )
    return merit_score




features = [
    ['x1', 'x2'],
    ['x2', 'x3'],
    ['x1', 'x3']
]
target = 'y'
summary = []
for subset in features:
    k = len(subset)
    r_cf = corr.loc[subset, target].mean()
    if k == 1:
        r_ff = 0
    else:
        feature_corr = corr.loc[subset, subset].values
        r_ff = (feature_corr.sum() - k) / (k * (k - 1))

    merit_score = (k * r_cf) / np.sqrt(
        k + k * (k - 1) * r_ff
    )
    summary.append({
        'Features': ', '.join(subset),
        'k': k,
        'Rcf': r_cf,
        'Rff': r_ff,
        'Merit Score': merit_score
    })
summary_df = pd.DataFrame(summary)
summary_df[['Rcf', 'Rff', 'Merit Score']] = \
    summary_df[['Rcf', 'Rff', 'Merit Score']].round(4)

display(summary_df)
```
