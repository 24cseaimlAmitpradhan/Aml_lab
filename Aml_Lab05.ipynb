```python
import numpy as np

x = np.array([1, 2, 3, 4], dtype=float)
y = np.array([2, 2, 4, 5], dtype=float)


n = len(x)
iterations = 2
alpha = 0.05
lam = 0.2

w = 0.0
b = 0.0

print("x values:", x)
print("y values:", y)
print("length of x:", n)

print("\nInitial weight (w):", w)
print("Initial bias (b):", b)


def predict(x, w, b):
    return w * x + b



for i in range(iterations):


    y_pred = predict(x, w, b)


    error = y_pred - y


    cost = (1 / (2 * n)) * np.sum(error ** 2) + (lam / 2) * w ** 2


    dw = (1 / n) * np.sum(error * x) + lam * w
    db = (1 / n) * np.sum(error)


    w = w - alpha * dw
    b = b - alpha * db


    new_prediction = predict(x, w, b)


    print("\nIteration:", i + 1)
    print("w:", w)
    print("b:", b)
    print("cost:", cost)
    print("New prediction:", new_prediction)



y_pred = predict(x, w, b)

print("\nFinal weight (w):", w)
print("Final bias (b):", b)
print("Final predictions:", y_pred)
print("Actual values:", y)
```
