```python
#  One R techinique
import pandas as pd
import numpy as np

data = pd.DataFrame({
    'Rainfall': ["High","High","High","Low","Low","Low"],
    'soiltype': ["Loamy","sundy","clay","Loamy","sundy","clay"],
    'yield':  ["Good","Good","Good","Poor","Poor","Good"]
})

def best_rule(data, feature, target):
    results = []

    for value in data[feature].unique():


        subset = data[data[feature] == value]

        majority_class = subset[target].mode()[0]

        prediction = np.where(
            data[feature] == value,
            majority_class,
            data[target].mode()[0]
        )

        accuracy = np.mean(prediction == data[target])

        results.append({
            'Feature': feature,
            'Rule': f"{feature} == '{value}'",
            'Prediction': majority_class,
            'Accuracy': accuracy
        })

    return pd.DataFrame(results)


rainfall_rules = best_rule(data, 'Rainfall', 'yield')
soiltype_rules = best_rule(data, 'soiltype', 'yield')

all_rules = pd.concat(
    [rainfall_rules, soiltype_rules],
    ignore_index=True
)

print("All rules:")
print(all_rules)


best = all_rules.loc[all_rules['Accuracy'].idxmax()]

print("\nBest single rule:")
print("Feature  :", best['Feature'])
print("Rule     :", best['Rule'])
print("Predict  :", best['Prediction'])
print("Accuracy :", best['Accuracy'])


```
