```python
#chi-square test
from scipy.stats import chisquare
observed=[10,5,5,10]
expected=[7.5,7.5,7.5,7.5]
chi2,chi2cal=chisquare(observed,expected)
print("chi2 value is:",chi2)
print("chi2 calculated value is:",chi2cal)
if chi2cal<chi2:
    print("reject null hypothesis")
else:
    print("accept null hypothesis")
```

```python
observed = [
    [10 , 5],
    [5, 10]
]

rows = len(observed)
cols = len(observed[0])

row_totals = []
for row in observed:
    row_totals.append(sum(row))

col_totals = []
for j in range(cols):
    total = 0
    for i in range(rows):
        total += observed[i][j]
    col_totals.append(total)

grand_total = sum(row_totals)

expected = []

for i in range(rows):
    row = []
    for j in range(cols):
        e = (row_totals[i] * col_totals[j]) / grand_total
        row.append(e)
    expected.append(row)

chi_square = 0

for i in range(rows):
    for j in range(cols):
        chi_square += ((observed[i][j] - expected[i][j]) ** 2) / expected[i][j]

df = (rows - 1) * (cols - 1)

chi_square_tabulated = 3.841

print("Observed Frequencies:")
for row in observed:
    print(row)

print("\nExpected Frequencies:")
for row in expected:
    print([round(x, 2) for x in row])

print("\nChi-Square Calculated =", round(chi_square, 3))
print("Degrees of Freedom =", df)
print("Chi-Square Tabulated =", chi_square_tabulated)

if chi_square > chi_square_tabulated:
    print("\n Reject H0 and  There is a significant association between the two variables.")
else:
    print("\nFail to Reject H0 and There is no significant association between the two variables.")

```
