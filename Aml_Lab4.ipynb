```python
# Linear Regression with Gradient Descent

import numpy as np
import matplotlib.pyplot as plt

# Dataset
x = np.array([1, 2, 3])
y = np.array([2, 3, 4])

n = len(x)
iteration = 2

print("x values:", x)
print("y values:", y)
print("Length of x:", n)

# Initial values
w = 0.0
b = 0.0
alpha = 0.1

print("Initial weight (w):", w)
print("Initial bias (b):", b)

# Prediction function
def predict(x, w, b):
    return w * x + b

# Cost function
def compute_cost(y, y_pred):
    return np.sum((y - y_pred) ** 2) / n

# Initial prediction and cost
y_pred = predict(x, w, b)
cost = compute_cost(y, y_pred)

print("\nInitial prediction:", y_pred)
print("Initial cost:", cost)

# Store cost values
cost_history = []

# Gradient Descent
for i in range(iteration):

    y_pred = predict(x, w, b)

    cost = compute_cost(y, y_pred)
    cost_history.append(cost)

    print("\nIteration:", i + 1)
    print("Old weight:", w)
    print("Old bias:", b)
    print("Prediction:", y_pred)
    print("Old cost:", cost)

    # Gradients
    dm = -2 / n * np.sum(x * (y - y_pred))
    db = -2 / n * np.sum(y - y_pred)

    print("dm:", dm)
    print("db:", db)

    # Update weight and bias
    w = w - alpha * dm
    b = b - alpha * db

    print("Updated weight:", w)
    print("Updated bias:", b)

    # New prediction and cost
    y_pred_new = predict(x, w, b)
    new_cost = compute_cost(y, y_pred_new)

    print("New prediction:", y_pred_new)
    print("New cost:", new_cost)

# Final results
print("\nFinal weight:", w)
print("Final bias:", b)
print("Final prediction:", predict(x, w, b))
print("Final cost:", compute_cost(y, predict(x, w, b)))

# -----------------------------------------
# Plot 1: Actual Data and Regression Line
# -----------------------------------------

plt.figure(figsize=(7, 5))

plt.scatter(x, y, label='Original Data')
plt.plot(x, predict(x, w, b), color='red', label='Linear Regression')

plt.xlabel('X')
plt.ylabel('Y')
plt.title('Linear Regression')
plt.legend()
plt.grid(True)

plt.show()

# -----------------------------------------
# Plot 2: Cost Function vs Iterations
# -----------------------------------------

plt.figure(figsize=(7, 5))

plt.plot(
    range(1, iteration + 1),
    cost_history,
    marker='o'
)

plt.xlabel('Iteration')
plt.ylabel('Cost')
plt.title('Cost Function vs Iterations')
plt.grid(True)

plt.show()













```
